```python
from pctoolkit.compressors import PromptCompressor
from pctoolkit.datasets import load_dataset
import pandas as pd
from tqdm import tqdm

compressor = PromptCompressor(
    type='LLMLingua2',
    device='cuda'
)

dataset_name = 'GSM'
dataset = load_dataset(dataset_name)

print("-----First-----")
print(dataset.data[0])

print("-----Second-----")
print(dataset.data[1])

ratio = 0.3
results = []

for sample in tqdm(dataset.data):
    question = sample["question"]
    answer = sample["answer"]
    compressed = compressor.compressgo(
            question,
            ratio)
    results.append({
        "original_question": question,
        "compressed_question": compressed,
        "answer": answer
    })
```

```python
df = pd.DataFrame(results)

print(df.head())

df.to_csv("gsm_compressed.csv", index=False)
```
